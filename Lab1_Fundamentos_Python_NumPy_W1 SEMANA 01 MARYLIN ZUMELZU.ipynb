
## Pregunta 1 (Listas) — 15 pts

Implementa la función `sumar_pares(nums)` que recibe una lista de enteros y retorna la **suma** de los números **pares**.  
Si la lista está vacía, retorna `0`.

```python
# TODO: implementa sumar_pares
def sumar_pares(nums):
    suma = 0
    # your code here
    for numero in nums:
        if numero % 2 == 0:
            suma = suma + numero
    return suma
```


## Pregunta 2 (Diccionarios) — 15 pts

Implementa `actualizar_contador(d, clave)` que:
- Si `clave` **existe** en el diccionario `d`, incrementa su valor en 1.  
- Si `clave` **no existe**, inicializa `d[clave] = 1`.  
La función retorna el diccionario modificado.

```python
# TODO: implementa actualizar_contador
def actualizar_contador(d, clave):
    # your code here
    if clave in d:
        d[clave] = d[clave] + 1
    else:
        d[clave] = 1
    return d
```

```python
d = {"a": 2}
out = actualizar_contador(d, "a")
assert out["a"] == 3 and d is out
```


## Pregunta 3 (Control de flujo) — 15 pts

Implementa `fizz_buzz(n)` que retorne:
- `"FizzBuzz"` si `n` es múltiplo de 3 **y** de 5,
- `"Fizz"` si solo es múltiplo de 3,
- `"Buzz"` si solo es múltiplo de 5,
- en otro caso, retorna el propio número `n`.

```python
# TODO: implementa fizz_buzz
def fizz_buzz(n):
    # your code here
    if n % 3 == 0 and n % 5 == 0:
        return "FizzBuzz"
    elif n % 3 == 0:
        return "Fizz"
    elif n % 5 == 0:
        return "Buzz"
    else:
        return n
```

```python
assert fizz_buzz(3) == "Fizz"
assert fizz_buzz(8) == 8
```

Implementa `calcular_promedio(nums)` que retorna el **promedio** de una lista de números.  
- Si la lista es vacía, retorna `0.0`.
- Debe funcionar con `int` o `float`.

```python
# TODO: implementa calcular_promedio
def calcular_promedio(nums):
     # your code here
    if len(nums) == 0:
        return 0.0
    return sum(nums) / len(nums)
```


## Pregunta 5 (NumPy: vectorización) — 20 pts

Escribe `zscore(x)` que reciba un `np.ndarray` 1D y retorne su **estandarización Z-score**:  
\[ z = (x - mean(x)) / std(x) \]
- Usa únicamente operaciones vectorizadas de NumPy (sin bucles `for`/`while`).
- Para `std == 0`, retorna un arreglo de ceros del mismo tamaño.

```python
# TODO: implementa zscore
def zscore(x):
    # your code here
    media = np.mean(x)
    desviacion = np.std(x)
    return (x - media) / desviacion
```

```python
arr = np.array([1,2,3,4,5], dtype=float)
z = zscore(arr)
assert z.shape == (5,)
assert abs(z.mean()) < 1e-7
assert abs(z.std() - 1.0) < 1e-7
```

## Pregunta 6 (NumPy: slicing & broadcasting) — 20 pts

Implementa `filtrar_y_escalar(m, umbral, k)` que:
- Recibe `m` como arreglo 2D `np.ndarray`.
- Retorna un **nuevo** arreglo donde todos los valores **> umbral** se multiplican por `k` (usa **filtrado booleano** y **broadcasting**).  
- No modifiques el arreglo `m` original.

```python
# TODO: implementa filtrar_y_escalar
def filtrar_y_escalar(m, umbral, k):
    # your code here
    resultado = m.copy()
    resultado[resultado > umbral] = resultado[resultado > umbral] * k
    return resultado
```

```python
m = np.array([[1,5],[10,2]], dtype=float)
out = filtrar_y_escalar(m, umbral=4, k=2)
assert np.allclose(out, np.array([[1,10],[20,2]], dtype=float))
# m no debe modificarse
assert np.allclose(m, np.array([[1,5],[10,2]], dtype=float))
```
