Para las visualizaciones, deberás crear funciones que reciban y retornen un `Axes` de `matplotlib`. A continuación un pequeño ejemplo: 

Vamos a visualizar una serie temporal de ventas muy básica: el total de unidades vendidas por día a partir de las columnas `fecha` y `unidades`. La función recibe un `Axes`, dibuja la curva agregada diaria y retorna ese mismo `Axes`.

```python
def grafico_intro(ax, df):
    ax.figure.set_size_inches(12, 4) # Opcional, para que se vea bien. 
    s = (df
         .set_index("fecha")["unidades"]
         .resample("D")
         .sum())

    ax.plot(s.index, s.values, label="Unidades/día")
    ax.set_title("Intro: recibir, modificar y retornar Axes")
    ax.set_xlabel("Fecha")
    ax.set_ylabel("Unidades")
    ax.legend()
    return ax
```

```python
ax = _new_ax()
ax = grafico_intro(ax, df) 
```
